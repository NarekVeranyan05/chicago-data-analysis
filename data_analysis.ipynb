# 1 Crimes Dataset

### DATA RETRIEVAL AND NULL COUNTS

```python
crime_df = pd.read_csv('./Chicago_2025_raw/Crimes_-_2001_to_Present_20260929.csv')
print("Number of rows: ", len(crime_df))
print("Number of columns: ", len(crime_df.columns))
crime_df.head()
```

```python
# number of NULL values in each column
crime_df.isnull().sum()
```

```python
# How many rows have a NULL for at least one column

print("Number of rows in the Crimes dataset that have a NULL: ", crime_df.isnull().any(axis=1).sum())
```

### DROPPING ROWS / COLUMNS

Attributes we drop:
* IURC
* Arrest
* X Coordinate
* Y Coordinate
* Year
* Location

How to deal with NULLs?
* for Location Description, leave as is
* for Ward, there's only one row – remove it
* for Community Area, remove those rows
* for (Latitude, Longitude), remove those rows

```python
# Dropping columns
crime_df = crime_df.drop(columns=['IUCR', 'Arrest', 'X Coordinate', 'Y Coordinate', 'Year', 'Location'])
# Dropping rows
crime_df = crime_df.dropna(subset=['Ward', 'Community Area', 'Latitude', 'Longitude'])

print("New number of rows: ", len(crime_df))
print("New number of columns: ", len(crime_df.columns))
crime_df.isnull().sum()
```

```python
crime_df.to_csv('Chicago_2025_cleaned/crimes.csv', index=False)
```

# 2 Arrests Dataset

### DATA RETRIEVAL AND NULL COUNTS

```python
arrests_df = pd.read_csv('./Chicago_2025_raw/Arrests_20260929.csv')
print("Number of rows: ", len(arrests_df))
print("Number of columns: ", len(arrests_df.columns))
arrests_df.head()
```

```python
# number of NULL values in each column
# We ignore individual CHARGE 1-4 CLASS, TYPE, STATUTE, and DESCRIPTION 
# because this will be modelled as subclasses.

arrests_df.drop(columns=[
    'CHARGE 1 STATUTE', 'CHARGE 1 DESCRIPTION', 'CHARGE 1 TYPE', 'CHARGE 1 CLASS',
    'CHARGE 2 STATUTE', 'CHARGE 2 DESCRIPTION', 'CHARGE 2 TYPE', 'CHARGE 2 CLASS',
    'CHARGE 3 STATUTE', 'CHARGE 3 DESCRIPTION', 'CHARGE 3 TYPE', 'CHARGE 3 CLASS',
    'CHARGE 4 STATUTE', 'CHARGE 4 DESCRIPTION', 'CHARGE 4 TYPE', 'CHARGE 4 CLASS'
]).isnull().sum()
```

```python
# How many rows have a NULL for at least one column
# We ignore individual CHARGE 1-4 CLASS, TYPE, STATUTE, and DESCRIPTION 
# because this will be modelled as subclasses.

print("Number of rows in the Arrests dataset that have a NULL (except CHARGE 1-4): ", 
      arrests_df.drop(columns=[
          'CHARGE 1 STATUTE', 'CHARGE 1 DESCRIPTION', 'CHARGE 1 TYPE', 'CHARGE 1 CLASS',
          'CHARGE 2 STATUTE', 'CHARGE 2 DESCRIPTION', 'CHARGE 2 TYPE', 'CHARGE 2 CLASS',
          'CHARGE 3 STATUTE', 'CHARGE 3 DESCRIPTION', 'CHARGE 3 TYPE', 'CHARGE 3 CLASS',
          'CHARGE 4 STATUTE', 'CHARGE 4 DESCRIPTION', 'CHARGE 4 TYPE', 'CHARGE 4 CLASS'
          ])
      .isnull().any(axis=1).sum()
    )
```

And all of this is due to CASE NUMBER. A person might get arrested for an incident that
was not reported. These are valid arrests, but irrelevant in the context of data analysis.
We will drop such rows.

The columns we drop are, for x = 1 to 4: CHARGE x STATUTE, CHARGE x DESCRIPTION, CHARGE x TYPE, CHARGE x CLASS.

This information about charge levels can be found from the CHARGES STATUTE, CHARGES DESCRIPTION,
CHARGES TYPE, CHARGES CLASS.

### DROPPING ROWS / COLUMNS

```python
arrests_df = arrests_df.drop(columns=[
    'CHARGE 1 STATUTE', 'CHARGE 1 DESCRIPTION', 'CHARGE 1 TYPE', 'CHARGE 1 CLASS',
    'CHARGE 2 STATUTE', 'CHARGE 2 DESCRIPTION', 'CHARGE 2 TYPE', 'CHARGE 2 CLASS',
    'CHARGE 3 STATUTE', 'CHARGE 3 DESCRIPTION', 'CHARGE 3 TYPE', 'CHARGE 3 CLASS',
    'CHARGE 4 STATUTE', 'CHARGE 4 DESCRIPTION', 'CHARGE 4 TYPE', 'CHARGE 4 CLASS'
])

arrests_df = arrests_df.dropna(subset=['CASE NUMBER'])

print(
    'Number of rows that have a NULL',
    arrests_df.isnull().any(axis=1).sum()
)

print('New number of rows', len(arrests_df))
print('New number of columns', len(arrests_df.columns))
```

```python
arrests_df.to_csv('Chicago_2025_cleaned/arrests.csv', index=False)
```
